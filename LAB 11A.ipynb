```python
import queue

# Create a LIFO Queue
lifo_queue = queue.LifoQueue()

# Add items to the queue
lifo_queue.put("PRODUCT 1")
lifo_queue.put("PRODUCT 2")
lifo_queue.put("PRODUCT 3")

# Display the items as they are removed from the queue
print("Removing items from LIFO Queue:")
while not lifo_queue.empty():
    print(lifo_queue.get())
```

```python
import queue

# Initialize a max-priority queue
priority_queue = queue.PriorityQueue()

# List of elements to be inserted into the queue
elements = [4, 8, 1, 7, 3]

# Insert elements into the priority queue with inverted priority
for element in elements:
    priority_queue.put(-element)  # Negate to mimic max-priority behavior

# Display the elements in max-priority order
print("Max-Priority Queue elements (in order of removal):")
while not priority_queue.empty():
    # Negate again to get the original values
    print(-priority_queue.get(), end=" ")
```

```python
class Vertex:
    def __init__(self, value):
        self.value = value
        self.neighbors = []

    def add_neighbor(self, neighbor):
        if neighbor not in self.neighbors:
            self.neighbors.append(neighbor)
            self.neighbors.sort()


class Graph:
    def __init__(self):
        self.vertices = {}

    def add_vertex(self, value):
        if value not in self.vertices:
            self.vertices[value] = Vertex(value)
        else:
            print(f"Vertex {value} already exists.")

    def add_edge(self, from_vertex, to_vertex):
        if from_vertex in self.vertices and to_vertex in self.vertices:
            self.vertices[from_vertex].add_neighbor(to_vertex)
            self.vertices[to_vertex].add_neighbor(from_vertex)
        else:
            print(f"One or both vertices {from_vertex}, {to_vertex} do not exist.")

    def add_vertices(self, values):
        for value in values:
            self.add_vertex(value)

    def display_graph(self):
        for key in sorted(self.vertices.keys()):
            print(f"{key}: {self.vertices[key].neighbors}")


# Example usage:
graph = Graph()

# Adding vertices
graph.add_vertices(['A', 'B', 'C', 'D', 'E'])

# Adding edges
graph.add_edge('A', 'B')
graph.add_edge('A', 'C')
graph.add_edge('B', 'C')
graph.add_edge('B', 'D')
graph.add_edge('C', 'E')

# Display the graph
graph.display_graph()
```
