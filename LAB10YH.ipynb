```python
# Original list of dinner guests
dinner_guests = ["ALI", "SHAHID", "AYESHA"]

# Announce that one guest can't make it
unavailable_guest = dinner_guests[1]
print(f"Unfortunately, {unavailable_guest} can't make it to dinner.")

# Replace the unavailable guest with a new guest
dinner_guests[1] = "HINA SHAIKH"

# Send out the new set of invitations
for guest in dinner_guests:
    print(f"Dear {guest}, you are cordially invited to dinner. I would be honored by your presence!")
```

```python
# Print a second set of invitation messages
print("\nSending out the second set of invitations:")
for guest in dinner_guests:
    print(f"Dear {guest}, you are still cordially invited to dinner. I look forward to your presence!")
```

```python
class Employee:
    # Class variable to keep track of total employees
    total_employees = 0

    def __init__(self, emp_name, salary):
        # Instance variables
        self.emp_name = emp_name
        self.salary = salary
        # Increment total employee count
        Employee.total_employees += 1

    @staticmethod
    def displayCount():
        """Display the total number of employees."""
        print(f"Total Employees: {Employee.total_employees}")

    def displayEmployee(self):
        """Display the employee's name and salary."""
        print(f"Employee Name: {self.emp_name}, Salary: {self.salary}")


# Example Usage
# Creating Employee instances
emp1 = Employee("ASFA NAVEED", 630000)
emp2 = Employee("AYZA ASIF", 340000)

# Display employee details
emp1.displayEmployee()
emp2.displayEmployee()

# Display total employee count
Employee.displayCount()
```
